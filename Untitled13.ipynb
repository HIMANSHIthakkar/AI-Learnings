```python
df = df.rename(columns={
    "Passage": "passage",
    "Questions": "question"
})
```

```python
df = df.groupby("passage")["question"].apply(
    lambda x: " ||| ".join(x)
).reset_index()

print("Dataset after grouping:", len(df))
df.head()
```

```python
print(len(df))
df.head(2)
```

```python
import torch
import pandas as pd
from transformers import MT5Tokenizer, MT5ForConditionalGeneration
```

```python
from transformers import MT5Tokenizer, MT5ForConditionalGeneration
import torch

model_name = "google/mt5-small"

tokenizer = MT5Tokenizer.from_pretrained(model_name)
model = MT5ForConditionalGeneration.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print("Tokenizer & Model Loaded ✅")
```

```python
def preprocess(examples):
    inputs = ["hi_IN प्रश्न निर्माण: " + p for p in examples["passage"]]
    targets = examples["question"]

    model_inputs = tokenizer(
        inputs,
        max_length=512,
        truncation=True,
        padding="max_length"
    )

    labels = tokenizer(
        targets,
        max_length=128,
        truncation=True,
        padding="max_length"
    )

    model_inputs["labels"] = labels["input_ids"]

    return model_inputs
```

```python
from datasets import Dataset

dataset = Dataset.from_pandas(df)
dataset = dataset.map(preprocess, batched=True)
dataset = dataset.train_test_split(test_size=0.1)

print("Dataset ready ✅")
```

```python
from transformers import Seq2SeqTrainingArguments

training_args = Seq2SeqTrainingArguments(
    output_dir="./mt5-qg",
    eval_strategy="epoch",        # evaluation each epoch
    learning_rate=3e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=3,
    weight_decay=0.01,
    save_total_limit=2,
    predict_with_generate=True,
    generation_max_length=128,
    generation_num_beams=5,
    logging_dir="./logs",
    fp16=True,
    report_to="none"
)

print("Training arguments ready ✅")
```

```python
from transformers import Seq2SeqTrainer

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    tokenizer=tokenizer
)

print("Trainer ready ✅")
```

```python
trainer.train()
```

```python
results = trainer.evaluate()
print(results)
```

```python
def generate_questions(passage):
    prompt = "hi_IN प्रश्न निर्माण: " + passage

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    output = model.generate(
        **inputs,
        max_length=128,
        num_beams=5
    )

    text = tokenizer.decode(output[0], skip_special_tokens=True)
    return text.split("|||")
```

```python
questions = generate_questions(my_passage)

for i, q in enumerate(questions, 1):
    print(f"{i}. {q.strip()}")
```
